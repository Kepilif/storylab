```python
import requests
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# connection
url = URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password="powodzenia",
    host="localhost",
    port=5432,
    database="international_debt",
)
engine = create_engine(url)

# API WB
indicator = "GC.DOD.TOTL.GD.ZS"
country = "POL"

#api_url = f"https://api.worldbank.org/v2/country/{country}/indicator/{indicator}?format=json&per_page=1000"
api_url = f"https://api.worldbank.org/v2/country/all/indicator/{indicator}?format=json&per_page=20000"
data = requests.get(api_url).json()[1]

# DataFrame
df = pd.DataFrame([{
    "country": r["country"]["value"],
    "country_code": r["country"]["id"],
    "year": int(r["date"]),
    "value": r["value"]
} for r in data if r["value"] is not None])

# zapis do Postgres
df.to_sql("wb_poland_public_debt", engine, if_exists="replace", index=False)

print("Zapisano:", len(df), "rekordów")
```
